## Dirichlet loss

The dirichlet loss can be derived form [Dirichlet-multinomial log-likelihood function](https://pmc.ncbi.nlm.nih.gov/articles/PMC4081639/).


Let the observed vote counts for an input sample be

$$
\mathbf{k} = (k_1, k_2, \ldots, k_K),  \qquad  n = \sum_{i=1}^{K} k_i
$$

Now , let the parameters of the Dirichlet distribution be
$$
\mathbf{\alpha} = (\alpha_1, \alpha_2, \ldots, \alpha_K) \qquad S= \sum_{i=1}^{K} \alpha_i

$$
---
A multinominal distribution is given by:

$$P(\mathbf{k} \mid \mathbf{p}) = \frac{\Gamma(n+1)}{\prod_i \Gamma(k_i + 1)} \prod_i p_i^{k_i}
$$ *where k is the voting counts*

Wheres a dirichlet distribution is given by:

$$
P(\mathbf{p} \mid \boldsymbol{\alpha}) = \frac{\Gamma(S)}{\prod_i \Gamma(\alpha_i)} \prod_i p_i^{\alpha_i - 1}
$$

### Integration and combining
$$
\mathbf{p} \sim \text{Multinominal Dirichlet}(\boldsymbol{\alpha,k})
\\
P(\mathbf{k} \mid \boldsymbol{\alpha}) = \int P(\mathbf{k} \mid \mathbf{p})\, P(\mathbf{p} \mid \boldsymbol{\alpha}) \, d\mathbf{p}
$$

This gives us 
$$
P(\mathbf{k} \mid \boldsymbol{\alpha}) = \frac{n!}{\prod_i k_i!} \, \frac{\Gamma(S)}{\Gamma(n+S)} \prod_i \frac{\Gamma(k_i+\alpha_i)}{\Gamma(\alpha_i)}
$$

### Taking negative log-likelihood


$$
\mathcal{L}(\boldsymbol{\alpha}) = -\log\Gamma(n+1) + \sum_i \log\Gamma(k_i+1) - \log\Gamma(S) + \log\Gamma(n+S) - \sum_i\Big[\log\Gamma(k_i+\alpha_i) - \log\Gamma(\alpha_i)\Big]
$$

since the backpropagation depends on the parameters $\alpha_i$, we can ignore the first two terms and we can write the loss function as:
$$
\mathcal{L}(\boldsymbol{\alpha}) \propto \log\Gamma(n+S) - \log\Gamma(S) - \sum_i\Big[\log\Gamma(k_i+\alpha_i) - \log\Gamma(\alpha_i)\Big]
$$

```python
import torch

def dirichlet_multinominal_nll_loss(k:torch.Tensor, alpha:torch.Tensor)->torch.Tensor:
    
    n= torch.sum(k,dim=-1)#along batch
    S = torch.sum(alpha,dim=-1)

    l = (torch.lgamma(n+S) - torch.lgamma(S)- 
         torch.sum(torch.lgamma(k+alpha)- torch.lgamma(alpha),dim=-1) )

    return l
```
